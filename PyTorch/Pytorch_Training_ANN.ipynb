```python
x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,1:],df.iloc[:,0], test_size=0.2)
```

```python
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train=torch.from_numpy(x_train)
x_test=torch.from_numpy(x_test)
y_train=torch.from_numpy(y_train)
y_test=torch.from_numpy(y_test)
```

```python
class mysimpleneural():

  def __init__(self,X):
    self.weights=torch.rand(X.shape[1],1,dtype=torch.float64,requires_grad=True)
    self.bias=torch.zeros(1,dtype=torch.float64,requires_grad=True)


  def forward(self, X):
    z=torch.matmul(X,self.weights)+self.bias
    y_pred=torch.sigmoid(z)
    return y_pred


  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train * torch.log(y_pred) + (1 - y_train) * torch.log(1 - y_pred)).mean()
    return loss
```

```python
learning_rate=0.1
epochs=25
```

```python
model = mysimpleneural(x_train)

for epoch in range(epochs): # Fixed spelling of epoch
    y_pred = model.forward(x_train)

    loss = model.loss_function(y_pred, y_train)
    loss.backward()

    with torch.no_grad():
        # Update parameters
        model.weights -= learning_rate * model.weights.grad
        model.bias -= learning_rate * model.bias.grad

        # Correct way to zero gradients manually
        model.weights.grad.zero_()
        model.bias.grad.zero_()

    # Fixed .iten() to .item()
    print(f"Epoch is {epoch+1} and the loss = {loss.item():.4f}")
```

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(x_test)
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == y_test).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
