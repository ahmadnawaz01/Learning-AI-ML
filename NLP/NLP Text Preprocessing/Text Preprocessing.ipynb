```python
import re
def removehtmltags(text):
    pattern=re.compile('<.*?>')
    return pattern.sub(r'',text)
```

# remove punctuation marks

```python
import string,time
```

```python
string.punctuation
```

```python
ex=string.punctuation
```

```python
def removepun(text):
    for char in ex:
        text=text.replace(char,'')
    return text
```

```python
text='ahsdjkes * ? ! $ali ahmad'
```

```python
removepun(text)
```

```python
start=time.time()
print(removepun(text))
tim1=time.time()-start
print(tim1)
```

```python
def removep(text):
    return text.translate(str.maketrans('','',ex))
```

# chat word treatment

```python
slang_dict = {}

with open('slang.txt', 'r') as file:
    for line in file:
        if line.strip():
            key, value = line.split('=', 1)
            slang_dict[key.strip()] = value.strip()

print(slang_dict)
```

```python
def chatconversion(text):
    new_text=[]
    for w in text.split():
        if w.upper() in slang_dict:
            new_text.append(slang_dict[w.upper()])
        else:
            new_text.append(w)
    return ' '.join(new_text)
```

```python
chatconversion('lmao i am good')
```

```python
from nltk.corpus import stopwords

def revestop(text):
    newtxt = []
    stop_words = set(stopwords.words('english'))
    
    for word in text.split():
        if word.lower() not in stop_words: 
            newtxt.append(word)
            
    return ' '.join(newtxt)
```

```python
revestop('i am yours brother')
```

# handling emoji

```python
import re
def remove_emoji(text):
    emoji_pattern = re.compile("["
                               u"\U0001F600-\U0001F64F"  # emoticons
                               u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                               u"\U0001F680-\U0001F6FF"  # transport & map symbols
                               u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                               u"\U00002702-\U000027B0"
                               u"\U000024C2-\U0001F251"
                               "]+", flags=re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji('😀 this is very funny ha😞😲😲')
```

## Stemming

### for english porterstemmer and for other lang snowball stemmer

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps= PorterStemmer()
```

```python
def stem_words(text):
    return ' '.join([ps.stem(word) for word in text.split()])
```

```python
sample="walked like you are loving it"
stem_words(sample)
```

```python
sample='walked walks walking walkeed'
stem_words(sample)
```

# Lemmatization

```python
import nltk

nltk.download('wordnet')

nltk.download('omw-1.4')

nltk.download('punkt')
from nltk.stem import WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations="?:!.,;"
sentence_words = nltk.word_tokenize(sentence)
for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words
print("{0:20}{1:20}".format("Word", "Lemma"))
for word in sentence_words:
    print ("{0:20}{1:20}".format(word,wordnet_lemmatizer.lemmatize(word,pos='v')))
```
