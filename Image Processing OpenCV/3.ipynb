```python
import numpy as np
import cv2
```

```python
img=cv2.imread('images/fruit.png')
```

```python
cv2.imshow('window',img)
cv2.waitKey(0)
```

```python
img.shape
```

```python
img=np.zeros((512,512,3))
img=(img * 255).astype(np.uint8)
```

```python
cv2.imshow('window',img)
cv2.waitKey(0)
```

```python
cv2.putText(img, text="hello", org=(100, 100), fontFace=cv2.FONT_ITALIC, fontScale=4, color=(0, 255, 255), thickness=2, lineType=cv2.LINE_AA)
```

```python
cv2.line(img,pt1=(0,0),pt2=(512,512),thickness=2,color=(0,255,0))
```

```python
#circle
cv2.circle(img,center=(100,400),radius=50,thickness=-1,color=(0,0,255))
```

```python
#rectangle
```

```python
cv2.rectangle(img,pt1=(100,100),pt2=(300,300),color=(255,0,0),thickness=-1)
```
